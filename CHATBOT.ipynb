```python
print("Simple Chatbot")
print("Type 'bye' to stop chatting")

while True:
    user = input("You: ").lower()

    if user == "hello":
        print("Bot: Hi! How are you?")

    elif user == "how are you":
        print("Bot: I am fine.")

    elif user == "your name":
        print("Bot: I am CodSoft Chatbot.")

    elif user == "bye":
        print("Bot: Goodbye!")
        break

    else:
        print("Bot: Sorry, I don't understand.")
```
