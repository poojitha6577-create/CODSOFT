```python
from IPython.display import Image, display

print("Simple Image Caption Generator")

image_name = input("Enter image name: ")

captions = {
    "cat": "A cute cat sitting on the floor.",
    "dog": "A happy dog playing outside.",
    "car": "A car parked on the road.",
    "tree": "A green tree in nature."
}

if image_name in captions:
    print("Caption:", captions[image_name])
else:
    print("Caption: Beautiful image.")
```
